# Pure Rotational Spectroscopy

In the **rigid rotor** model a diatomic molecule rotates about its center of mass
at a fixed bond length $R_e$. The rotational energy levels (in $\mathrm{cm}^{-1}$) are:

\begin{equation}
F(J) = B J(J+1), \quad J = 0, 1, 2, \ldots \tag{1}
\end{equation}

where $B$ is the **rotational constant**:

\begin{equation}
B = \frac{\hbar}{4\pi c I}, \quad I = \mu R_e^2, \quad \mu = \frac{m_1 m_2}{m_1 + m_2} \tag{2}
\end{equation}

with $c$ in cm $\mathrm{s}^{-1}$. The selection rule $\Delta J = \pm 1$ restricts absorption
transitions $J \rightarrow J+1$ to wavenumbers:

\begin{equation}
\tilde{\nu}(J \rightarrow J+1) = F(J+1) - F(J) = 2B(J+1) \tag{3}
\end{equation}

Only molecules with a permanent dipole moment show a pure rotational spectrum.
